```python
import networkx as nx
```

```python
g1 = nx.Graph() #Instanciamos una clase graph
```

```python
# Agregamos nodos a nuestro grafo

for i in range(7):
    g1.add_node(f'n{i+1}')

g1.nodes
```

```python
# Establezcamos ahora las conexiones

g1.add_edges_from([('n1','n7'),('n2','n3'),('n4','n6'),('n5','n1')])
```

```python
# Grafiquemos lo que hemos hecho

nx.draw(g1, with_labels = True, node_color = 'green', node_size = 1000, font_color = 'white')
```

```python
# Número de enlaces y nodos
g1.number_of_edges(), g1.number_of_nodes()
```

```python
# Si queremos ver los nodos adyacentes a uno podemos hacerlo con neighbors, pero este es un generador, por lo que tendriamos que imprimirlo

for i in g1.neighbors('n1'):
    print(i)
```

```python
g1.has_node('n1')
```

```python
g1.has_edge('n1','n7')
```

### Divisor de

```python
g_numeros = nx.Graph()
```

```python
for i in range(20):
    g_numeros.add_node(f'{i+1}')
```

```python
g_numeros.nodes
```

```python
for i in g_numeros:
    for j in g_numeros:
        if i != j:
            if int(i)%int(j) == 0:
                g_numeros.add_edge(i,j)
```

```python
nx.draw(g_numeros, with_labels=True, node_size=700, node_color="lightblue", font_weight="bold")
```

### Segundo ejemplo

```python
G2, G3 = nx.Graph(), nx.Graph()
G2, G3
```

```python
nodos1 = list(range(10))
print(nodos1)

nodos2 = list(range(10,100))
```

```python
G2.add_nodes_from(nodos1)
G3.add_nodes_from(nodos2)
```

```python
G3.nodes()
nx.draw(G2, with_labels = True)
```

```python
# Para agregar manualmente los enlaces podríamos definir una función

def generar_parejas(rango, n_parejas):
    from random import sample
    parejas = []
    for i in range(n_parejas):
        pareja = tuple(sample(range(rango),2))
        parejas.append(pareja)
    return parejas
```

```python
# Generemos pares aleatorios

enlace1 = generar_parejas(10,5)
print(enlace1)
```

```python
# Agreguemos los enclaes al grafo

G2.add_edges_from(enlace1)
```

```python
G2.edges
```

```python
nx.draw(G2, with_labels = True)
```

```python
nx.is_connected(G2) # No está conectada
```

```python
nx.has_path(G2,3,4) # Pregunto si hay camino entre las graficas
```

```python
# Contemos loc omponentes
nx.number_of_edges(G2) # Número de enlaces
```

```python
nx.number_connected_components(G2) # 6 componentes o agrupaciones, o subgrafos
```

```python
enlace2 = generar_parejas(100,100)
G3.add_edges_from(enlace2)

nx.draw(G3, with_labels = True)
```

```python
nx.is_tree(G2) #pregunta si se trata de un arbol
```

```python
for i in nx.connected_components(G2):
    print(i)
```
