```python
import networkx as nx
```

```python
g1 = nx.Graph() #Instanciamos una clase graph
```

```python
# Agregamos nodos a nuestro grafo

for i in range(7):
    g1.add_node(f'n{i+1}')

g1.nodes
```

```python
# Establezcamos ahora las conexiones

g1.add_edges_from([('n1','n7'),('n2','n3'),('n4','n6'),('n5','n1')])
```

```python
# Grafiquemos lo que hemos hecho

nx.draw(g1, with_labels = True, node_color = 'green', node_size = 1000, font_color = 'white')
```

```python
# Número de enlaces y nodos
g1.number_of_edges(), g1.number_of_nodes()
```

```python
# Si queremos ver los nodos adyacentes a uno podemos hacerlo con neighbors, pero este es un generador, por lo que tendriamos que imprimirlo

for i in g1.neighbors('n1'):
    print(i)
```

```python
g1.has_node('n1')
```

```python
g1.has_edge('n1','n7')
```

### Divisor de

```python
g_numeros = nx.Graph()
```

```python
for i in range(20):
    g_numeros.add_node(f'{i+1}')
```

```python
g_numeros.nodes
```

```python
for i in g_numeros:
    for j in g_numeros:
        if i != j:
            if int(i)%int(j) == 0:
                g_numeros.add_edge(i,j)
```

```python
nx.draw(g_numeros, with_labels=True, node_size=700, node_color="lightblue", font_weight="bold")
```
